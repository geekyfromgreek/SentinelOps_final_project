# Feature Engineering

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("/kaggle/input/datasets/stephanmatzka/predictive-maintenance-dataset-ai4i-2020/ai4i2020.csv")
```

# Feature Encoding
now, we need to clean the data and ENcoe it to make predictions for our model

```python
df=pd.get_dummies(df,columns=["Type"],drop_first=True)
```

```python
df
```

did Dummmy variable encoding because if L and M are 0, we know its H machine
--

# Removing identifiers


```python
df=df.drop(["UDI","Product ID"],axis=1)
```

identifiers arent needed and they can cause model to learn things its not supposed to know
--

 # Feature removal

```python
df=df.drop(["TWF","HDF","PWF","OSF","RNF"],axis=1)
```

```python
df
```

Failure types would describe the exact type of failire,but we need to predict it so we removed failure types
